Выделим четыре столбца: размер RAM, ROM, возможность установки дополнительной памяти и её объём. Объём будем измерять в ГБ.

Если дополнительную память установить нельзя введём значение 0. В данном случае корректно обработать данные нам поможет индикаторная переменная, которую мы добавляем. Без такой индикаторной переменной обучаемый вес для размера расширяемой памяти будет искажаться.

```python
def extract_RAM_ROM(x):
    parts = x.split("|")
    result = []
    scale_dict = {"TB":1024, "GB":1, "MB":1/1024, "KB": 1/1024/1024}
    # Извлекаем информацию про RAM
    ram = parts[0].strip().split(" ")
    scale = scale_dict[ram[1]]
    result.append(float(ram[0]) * scale)
    # Извлекаем информацию про ROM
    rom = parts[1].strip().split(" ")
    scale = scale_dict[rom[1]]
    result.append(float(rom[0]) * scale)
    
    if len(parts) == 3:
        external = parts[2].strip().split(" ")
        result.append(True)
        scale = scale_dict[external[-1]]
        result.append(float(external[-2]) * scale)
    else:
        result.append(False)
        result.append(0)
    return result
        
data = df["RAM & ROM"].apply(extract_RAM_ROM)
data
```

```python
data = data.apply(pd.Series)
df["ram"] = data[0]
df["rom"] = data[1]
df["is_extendable"] = data[2]
df["extension_size"] = data[3]
df.head()
```

```python
import numpy as np
import pandas as pd
import re
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.metrics import r2_score, root_mean_squared_error
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import make_pipeline

df = pd.read_csv("indian_smartphones_under_20000_raw.csv", index_col=0)
df.drop_duplicates(inplace=True)
df = df[df["Price"].notna()].copy()

data = df["RAM & ROM"].str.split("|")

data1 = data.apply(lambda x: x[0].strip().split(" "))
df = df[data1.apply(lambda x: len(x)) != 4]

data1 = df["RAM & ROM"].str.split("|").apply(lambda x: x[0].strip().split(" "))
words = data1.apply(pd.Series)
df = df[words[2] == "RAM"]

data2 = df["RAM & ROM"].str.split("|").apply(lambda x: x[1].strip().split(" "))
words2 = data2.apply(pd.Series)
df = df[data2.apply(lambda x: len(x)) != 4]
df = df[words2[2] == "ROM"]

data_battery = df["Battery Capacity"].str.split(" ")
words_battery = data_battery.apply(pd.Series)
df = df[words_battery[1] == "mAh"]

scale_dict = {"TB": 1024, "GB": 1, "MB": 1/1024, "KB": 1/1024/1024}

def extract_RAM_ROM(x):
    parts = x.split("|")
    result = []
    ram = parts[0].strip().split(" ")
    result.append(float(ram[0]) * scale_dict.get(ram[1], 1))
    rom = parts[1].strip().split(" ")
    result.append(float(rom[0]) * scale_dict.get(rom[1], 1))
    if len(parts) == 3:
        external = parts[2].strip().split(" ")
        result.append(True)
        result.append(float(external[-2]) * scale_dict.get(external[-1], 1))
    else:
        result.append(False)
        result.append(0)
    return result

data = df["RAM & ROM"].apply(extract_RAM_ROM).apply(pd.Series)
df["ram"] = data[0]
df["rom"] = data[1]
df["is_extendable"] = data[2].astype(int)
df["extension_size"] = data[3]

def parse_battery(text):
    return float(text.split(" ")[0])


df["battery_capacity"] = df["Battery Capacity"].apply(parse_battery)
df = df[df["battery_capacity"].notna()]

def extract_brand(name):
    name = str(name).strip()
    first_word = name.split(' ')[0]
    brand_map = {
        'vivo': 'vivo', 'MOTOROLA': 'Motorola', 'realme': 'realme',
        'Samsung': 'Samsung', 'Xiaomi': 'Xiaomi', 'Redmi': 'Xiaomi',
        'OPPO': 'OPPO', 'POCO': 'POCO', 'Tecno': 'Tecno',
        'Infinix': 'Infinix', 'Nokia': 'Nokia', 'OnePlus': 'OnePlus',
        'I': 'I Kall', 'Snexian': 'Snexian', 'LAVA': 'LAVA',
        'PHILIPS': 'Philips', 'Micromax': 'Micromax', 'HMD': 'HMD',
        'Apple': 'Apple', 'Google': 'Google', 'Nothing': 'Nothing'
    }
    return brand_map.get(first_word, first_word)

df['Brand'] = df['Model Name'].apply(extract_brand)

def extract_display_size(text):
    match = re.search(r'(\d+\.?\d*)\s*(?:cm|inch)', str(text))
    return float(match.group(1)) if match else np.nan

df['display_size_cm'] = df['Display Size'].apply(extract_display_size)

def extract_warranty_months(text):
    if pd.isna(text) or not isinstance(text, str):
        return np.nan
    text_lower = text.lower()
    if 'no warranty' in text_lower:
        return 0
    year_match = re.search(r'(\d+)\s*year', text_lower)
    if year_match:
        return int(year_match.group(1)) * 12
    month_match = re.search(r'(\d+)\s*month', text_lower)
    if month_match:
        return int(month_match.group(1))
    return np.nan

df['warranty_months'] = df['Warranty'].apply(extract_warranty_months)
df['warranty_missing'] = df['warranty_months'].isna().astype(int)
df['warranty_months'] = df['warranty_months'].fillna(0)

def normalize_processor(text):
    if pd.isna(text) or not isinstance(text, str):
        return 'Unknown'
    text = text.strip()
    text = re.sub(r'\s+Processor$', '', text, flags=re.IGNORECASE)
    text = re.sub(r'\s+\d+G$', '', text, flags=re.IGNORECASE)
    return text.strip()

df['processor'] = df['Processor'].apply(normalize_processor)

def parse_discount(discount_str):
    if pd.isna(discount_str) or not isinstance(discount_str, str):
        return np.nan
    match = re.search(r'(\d+)%', discount_str)
    return int(match.group(1)) / 100.0 if match else np.nan

df['discount_percent'] = df['Percentage Discount'].apply(parse_discount)
df['discount_percent'] = df['discount_percent'].fillna(0)

df['original_price'] = df['Price'] / (1 - df['discount_percent'])
df['original_price'] = df['original_price'].replace([np.inf, -np.inf], np.nan).fillna(df['Price'])

df['display_size_cm'] = df['display_size_cm'].fillna(df['display_size_cm'].median())

for col in ['ram', 'rom', 'extension_size', 'battery_capacity',
            'display_size_cm', 'warranty_months', 'discount_percent',
            'original_price']:
    df[col] = pd.to_numeric(df[col], errors='coerce')

features = ['ram', 'rom', 'is_extendable', 'extension_size', 'battery_capacity',
            'display_size_cm', 'warranty_months', 'warranty_missing',
            'Brand', 'processor', 'discount_percent']
target = 'original_price'

df_model = df[features + [target]].copy()
df_model = df_model.dropna()

X = df_model[features]
Y = df_model[target]

numerical_features = ['ram', 'rom', 'is_extendable', 'extension_size',
                      'battery_capacity', 'display_size_cm', 'warranty_months',
                      'warranty_missing', 'discount_percent']
categorical_features = ['Brand', 'processor']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_features),
        ('cat', OneHotEncoder(handle_unknown='ignore', min_frequency=3), categorical_features)
    ])

X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=42)

model = make_pipeline(preprocessor, Ridge(alpha=1.0))
model.fit(X_train, Y_train)

Y_pred = model.predict(X_test)
print("Результаты")
print(f"R2 Score: {r2_score(Y_test, Y_pred):.4f}")
print(f"RMSE: {root_mean_squared_error(Y_test, Y_pred):.2f}")

scores = cross_val_score(model, X, Y, cv=5, scoring='r2')
print(f"CV R2: {scores.mean():.4f} ± {scores.std():.4f}")
```
